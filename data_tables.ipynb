### 2.normalize to the catogry names.

```python
sub_category_map = {
    "TON": "Toner",
    "LJ": "Laserjet",
    "IJ": "Inkjet",
    "INK": "Ink",
    "SJ": "SJ",
    "UNK": "UNK",
    "Laptop": "Laptop",
    "Monitor": "Monitor",
    "Desktop": "Desktop"
}

pl_df["lpp_sub_category"] = pl_df["sub_category"].map(sub_category_map)

pl_df[["pl", "sub_category", "lpp_sub_category"]].head(10)
```

### 3.identify the brand

```python
pl_df["brand"] = pl_df["pl"].str.extract(r"^(DELL|HP)", expand=False)

pl_df[["pl", "brand"]].head(10)
```

### 4.create LPP Perccentage mapping.

```python
lpp_rates = {
    ("HP", "Laptop"): 0.034,
    ("DELL", "Laptop"): 0.078,

    ("HP", "Monitor"): 0.012,
    ("DELL", "Monitor"): 0.056,

    ("HP", "Toner"): 0.091,
    ("DELL", "Toner"): 0.023,

    ("HP", "Laserjet"): 0.065,
    ("DELL", "Laserjet"): 0.035,

    ("HP", "Inkjet"): 0.012,
    ("DELL", "Inkjet"): 0.056,

    ("HP", "Desktop"): 0.091,
    ("DELL", "Desktop"): 0.023,

    ("HP", "Ink"): 0.065,
    ("DELL", "Ink"): 0.035
}
```

```python
pl_df["lpp_percentage"] = [
    lpp_rates.get((brand, subcat))
    for brand, subcat in zip(
        pl_df["brand"],
        pl_df["lpp_sub_category"]
    )
]
```

```python
from docx import Document

template = "Violation_Email.docx"

def create_letter(row):
    doc = Document(template)

    for p in doc.paragraphs:
        for key, value in row.items():
            p.text = p.text.replace("{" + key + "}", str(value))

    for table in doc.tables:
        for cell in table._cells:
            for key, value in row.items():
                cell.text = cell.text.replace("{" + key + "}", str(value))

    file = f"Violation_{row['SKU']}_{row['Violation_count']}.docx"
    doc.save(file)
    return file
```
